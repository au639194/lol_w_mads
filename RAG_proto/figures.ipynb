```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
```

```python
# Adjust positions for better alignment and shorter arrows
positions_azure_updated = {
    'User Upload\n(Any Web Client)': (0.1, 0.8),
    'Web API\n(Flask/Django)': (0.3, 0.7),
    'Storage\n(Azure Blob Storage)': (0.5, 0.8),
    'Text Extraction\n(Azure Cognitive Services)': (0.3, 0.5),
    'Embedding Generation\n(Azure OpenAI - Ada)': (0.5, 0.6),
    'Vector Store\n(Azure Search/Vector DB)': (0.7, 0.7),
    'Duplicate Check\n(Azure Search)': (0.7, 0.5),
    'Task Queue\n(Azure Functions)': (0.3, 0.3),
    'Processed Results\n(Azure Cosmos DB/Blob)': (0.5, 0.3),
    'Monitoring\n(Azure Monitor)': (0.9, 0.8),
    'Logging\n(Azure Log Analytics)': (0.1, 0.6)
}

# Create figure and axis
fig, ax = plt.subplots(figsize=(12, 8))

# Add rectangles for components with updated labels
rectangles = {}
for label, (x, y) in positions_azure_updated.items():
    rect = patches.Rectangle((x - 0.05, y - 0.05), 0.1, 0.1, fill=True, color='lightblue', edgecolor='black')
    ax.add_patch(rect)
    rectangles[label] = rect
    ax.text(x, y, label, ha='center', va='center', fontsize=9, weight='bold')

# Add arrows between components with updated positions
arrows_azure_updated = [
    ('User Upload\n(Any Web Client)', 'Web API\n(Flask/Django)'),
    ('Web API\n(Flask/Django)', 'Storage\n(Azure Blob Storage)'),
    ('Storage\n(Azure Blob Storage)', 'Text Extraction\n(Azure Cognitive Services)'),
    ('Text Extraction\n(Azure Cognitive Services)', 'Embedding Generation\n(Azure OpenAI - Ada)'),
    ('Embedding Generation\n(Azure OpenAI - Ada)', 'Vector Store\n(Azure Search/Vector DB)'),
    ('Vector Store\n(Azure Search/Vector DB)', 'Duplicate Check\n(Azure Search)'),
    ('Duplicate Check\n(Azure Search)', 'Processed Results\n(Azure Cosmos DB/Blob)'),
    ('Web API\n(Flask/Django)', 'Task Queue\n(Azure Functions)'),
    ('Task Queue\n(Azure Functions)', 'Text Extraction\n(Azure Cognitive Services)'),
    ('Task Queue\n(Azure Functions)', 'Embedding Generation\n(Azure OpenAI - Ada)'),
    ('Monitoring\n(Azure Monitor)', 'Storage\n(Azure Blob Storage)'),
    ('Logging\n(Azure Log Analytics)', 'Task Queue\n(Azure Functions)'),
]

for start, end in arrows_azure_updated:
    start_rect = rectangles[start]
    end_rect = rectangles[end]
    
    # Calculate edge points for arrows
    start_x = start_rect.get_x() + start_rect.get_width() / 2
    start_y = start_rect.get_y() + start_rect.get_height() / 2
    end_x = end_rect.get_x() + end_rect.get_width() / 2
    end_y = end_rect.get_y() + end_rect.get_height() / 2
    
    # Adjust for arrow direction
    if start_x < end_x:
        start_x = start_rect.get_x() + start_rect.get_width()
        end_x = end_rect.get_x()
    elif start_x > end_x:
        start_x = start_rect.get_x()
        end_x = end_rect.get_x() + end_rect.get_width()
    
    if start_y < end_y:
        start_y = start_rect.get_y() + start_rect.get_height()
        end_y = end_rect.get_y()
    elif start_y > end_y:
        start_y = start_rect.get_y()
        end_y = end_rect.get_y() + end_rect.get_height()
    
    ax.annotate('', xy=(end_x, end_y), xytext=(start_x, start_y),
                arrowprops=dict(facecolor='black', arrowstyle='->'))

# Add title
ax.set_title('Optimized RAG Pipeline with Azure OpenAI: File Upload, Processing, and Duplicate Checking', fontsize=14, weight='bold')

# Hide axes
ax.axis('off')

# Show plot
plt.show()
```
